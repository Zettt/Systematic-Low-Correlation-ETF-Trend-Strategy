# Sector Rotation Low-Correlation ETF Strategy


```python
import sys
from pathlib import Path

import pandas as pd

sys.path.append(str(Path.cwd().parent))

from backtester import Backtester
from data_fetcher import fetch_strategy_data
from strategy_config import get_strategy
```

```python
strategy = get_strategy("sector_rotation")
if not strategy.daily_prices_path.exists() or not strategy.weekly_prices_path.exists():
    fetch_strategy_data(strategy)

daily_prices = pd.read_csv(strategy.daily_prices_path, parse_dates=["Date"])
weekly_prices = pd.read_csv(strategy.weekly_prices_path, parse_dates=["Date"])
daily_prices.head()
```

```python
bt = Backtester(daily_prices, weekly_prices, strategy=strategy)
bt.run_backtest(initial_capital=10000, rebalance_freq="M")
bt.get_performance_metrics()
```

```python
fig = bt.plot_results()
fig.show()
```
